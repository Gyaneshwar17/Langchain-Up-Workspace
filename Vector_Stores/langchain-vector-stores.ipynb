```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_classic.vectorstores import Chroma
```

```python
from langchain_classic.schema import Document

doc1 = Document(
    page_content="Max Verstappen is a dominant Formula 1 driver known for his aggressive racecraft, relentless pace, and multiple World Championships with Red Bull Racing.",
    metadata={"team": "Red Bull Racing", "role": "Driver"}
)

doc2 = Document(
    page_content="Lewis Hamilton is one of the most successful drivers in Formula 1 history, tied for a record seven World Championships and renowned for his wet-weather mastery and longevity.",
    metadata={"team": "Ferrari", "role": "Driver"}
)

doc3 = Document(
    page_content="Toto Wolff is the Team Principal and CEO of the Mercedes-AMG Petronas Formula One Team, famous for leading the team to eight consecutive Constructors' Championships.",
    metadata={"team": "Mercedes", "role": "Team Principal"}
)

doc4 = Document(
    page_content="Susie Wolff is a former racing driver and the Managing Director of F1 Academy, leading initiatives to develop and promote female talent in motorsport.",
    metadata={"team": "F1 Academy", "role": "Managing Director"}
)

doc5 = Document(
    page_content="Michael Schumacher is a Formula 1 legend who won seven World Championships, largely defining an era of dominance with Scuderia Ferrari in the early 2000s.",
    metadata={"team": "Ferrari", "role": "Driver"}
)

docs = [doc1, doc2, doc3, doc4, doc5]
```

```python

vector_store = Chroma(
    embedding_function=GoogleGenerativeAIEmbeddings(model='gemini-embedding-001',output_dimensionality=300),
    persist_directory='chroma_db',
    collection_name='sample'
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
vector_store.similarity_search(
    query='who amoung them is the best driver?',
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query="who is the best driver of all time?",
    k=1
)
```
